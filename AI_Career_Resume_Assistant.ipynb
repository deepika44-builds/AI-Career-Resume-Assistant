```python
!pip install -q google-genai

from google import genai
from google.colab import userdata
import traceback

try:
    print("STEP 1: Connecting to Gemini...")

    api_key = userdata.get("GEMINI_API_KEY")

    if not api_key:
        raise ValueError("GEMINI_API_KEY not found in Colab Secrets.")

    client = genai.Client(api_key=api_key)

    print("STEP 2: Checking resume text...")

    if "resume_text" not in globals() or not resume_text.strip():
        raise ValueError("Resume text not found. Please run the PDF upload and reading cell first.")

    print("STEP 3: Sending resume for analysis...")

    prompt = f"""
    You are an AI career advisor.

    Analyse the following student resume and provide:

    1. Overall resume feedback
    2. Strengths
    3. Technical skills identified
    4. Missing or useful skills to learn
    5. Suggestions to improve the resume
    6. Suitable entry-level internships and job roles
    7. A resume score out of 10 with reasons

    Be specific, constructive and beginner-friendly.
    Do not invent qualifications, skills or experience.

    RESUME:
    {resume_text[:12000]}
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    print("\n" + "=" * 50)
    print("AI CAREER & RESUME ASSISTANT")
    print("=" * 50)
    print(response.text)

except Exception as e:
    print("\nSomething went wrong:")
    print(str(e))
    traceback.print_exc()
```

```python
# JOB DESCRIPTION MATCHER

job_description = input("Paste the Job Description here:\n\n")

if not job_description.strip():
    print("Please enter a job description.")
else:
    prompt = f"""
    Compare this resume with the job description.

    RESUME:
    {resume_text[:12000]}

    JOB DESCRIPTION:
    {job_description[:8000]}

    Provide:
    1. Resume-to-job match score out of 100
    2. Matching skills
    3. Missing skills
    4. Relevant strengths
    5. Skills to improve
    6. Resume improvements for this role
    7. Interview preparation topics

    Be realistic. Do not invent skills or experience.
    """

    try:
        print("\nAnalysing your job match...\n")

        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        print("=" * 55)
        print("JOB DESCRIPTION MATCH REPORT")
        print("=" * 55)
        print(response.text)

    except Exception as e:
        print("Error:", e)
```

```python
# LOCAL AI INTERVIEW PREPARATION

target_role = input("Enter target job role: ").strip()

interview_questions = {
    "python": [
        "What are Python's main features?",
        "What is the difference between a list and a tuple?",
        "What are functions in Python?",
        "Explain loops in Python.",
        "What is exception handling?",
        "What is object-oriented programming?",
        "What are Python dictionaries?",
        "What is the purpose of the __init__ method?",
        "How do you read a file in Python?",
        "What is the difference between == and is?"
    ],
    "data": [
        "What is data analysis?",
        "What is NumPy used for?",
        "What is Pandas?",
        "What is data cleaning?",
        "What is the difference between a DataFrame and a Series?",
        "What is supervised learning?",
        "What is unsupervised learning?",
        "What is a classification model?",
        "What is overfitting?",
        "How do you handle missing data?"
    ],
    "software": [
        "What is a data structure?",
        "Explain the concept of OOP.",
        "What is an algorithm?",
        "What is version control?",
        "What is GitHub?",
        "What is debugging?",
        "What is an API?",
        "What is database management?",
        "Explain time complexity.",
        "What is software testing?"
    ]
}

role = target_role.lower()

if not target_role:
    print("Please enter a job role.")

else:
    if "data" in role or "machine learning" in role:
        questions = interview_questions["data"]
    elif "python" in role:
        questions = interview_questions["python"]
    else:
        questions = interview_questions["software"]

    print("\n" + "=" * 55)
    print("AI CAREER & RESUME ASSISTANT")
    print("INTERVIEW PREPARATION")
    print("=" * 55)

    print("\nTarget Role:", target_role)

    print("\nTECHNICAL INTERVIEW QUESTIONS")
    for i, question in enumerate(questions, 1):
        print(f"{i}. {question}")

    print("\nHR INTERVIEW QUESTIONS")
    hr_questions = [
        "Tell me about yourself.",
        "Why are you interested in this internship?",
        "What are your strengths?",
        "Describe a project you have worked on.",
        "How do you handle learning something new?"
    ]

    for i, question in enumerate(hr_questions, 1):
        print(f"{i}. {question}")

    print("\nPROJECT-BASED QUESTIONS")
    project_questions = [
        "Explain your project and its main objective.",
        "Which technologies did you use and why?",
        "What challenges did you face while developing it?"
    ]

    for i, question in enumerate(project_questions, 1):
        print(f"{i}. {question}")

    print("\n" + "=" * 55)
    print("INTERVIEW QUESTION GENERATION COMPLETED!")
    print("=" * 55)
```
